```python
# Preserve original timestamps, including when this cell is rerun.
if "transaction_datetime_original" not in df.columns:
    df["transaction_datetime_original"] = df["transaction_datetime"].copy()

original_dates = pd.to_datetime(
    df["transaction_datetime_original"],
    format="%Y-%m-%d %H:%M:%S",
    errors="raise",
)

# Replace only the year.
df["transaction_datetime"] = original_dates.map(
    lambda timestamp: timestamp.replace(year=2025)
    if pd.notna(timestamp)
    else pd.NaT
)

display(
    df[[
        "transaction_datetime_original",
        "transaction_datetime",
    ]].head()
)

print("Updated date range:")
print(df["transaction_datetime"].min())
print(df["transaction_datetime"].max())
```

```python
for column in ["transaction_day_of_week", "transaction_hour"]:
    original_column = f"{column}_original"

    if original_column not in df.columns:
        df[original_column] = df[column].copy()

df["transaction_day_of_week"] = (
    df["transaction_datetime"].dt.day_name()
)

df["transaction_hour"] = (
    df["transaction_datetime"].dt.hour.astype("Int64")
)

display(
    df[[
        "transaction_datetime",
        "transaction_day_of_week_original",
        "transaction_day_of_week",
        "transaction_hour",
    ]].head(10)
)
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import StrMethodFormatter

if df.empty:
    raise ValueError("Dataset is empty.")

fraud_target = pd.to_numeric(df["is_fraud"], errors="coerce")

if not fraud_target.isin([0, 1]).all():
    raise ValueError("is_fraud must contain only 0 and 1, with no missing values.")

fraud_summary = (
    fraud_target.value_counts()
    .reindex([0, 1], fill_value=0)
    .rename_axis("is_fraud")
    .reset_index(name="transaction_count")
)

fraud_summary["transaction_type"] = (
    fraud_summary["is_fraud"].map({0: "Legitimate", 1: "Fraud"})
)

fraud_summary["percentage"] = (
    fraud_summary["transaction_count"] / len(df) * 100
)

display(
    fraud_summary[[
        "transaction_type",
        "transaction_count",
        "percentage",
    ]].style.format({
        "transaction_count": "{:,.0f}",
        "percentage": "{:.3f}%",
    })
)

fig, ax = plt.subplots(figsize=(4, 5))

bars = ax.bar(
    fraud_summary["transaction_type"],
    fraud_summary["transaction_count"],
    color=["#14E629", "#DC2626"],
    width=0.55,
)

labels = [
    f"{count:,.0f}\n({percentage:.3f}%)"
    for count, percentage in zip(
        fraud_summary["transaction_count"],
        fraud_summary["percentage"],
    )
]

ax.bar_label(bars, labels=labels, padding=5)
ax.set_title("Transaction Distribution: Legitimate vs Fraud", pad=15)
ax.set_ylabel("Number of transactions")
ax.yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))
ax.set_ylim(0, fraud_summary["transaction_count"].max() * 1.18)
ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

amount_data = df[["transaction_amount", "is_fraud"]].copy()

amount_data["transaction_amount"] = pd.to_numeric(
    amount_data["transaction_amount"], errors="coerce"
)
amount_data["is_fraud"] = pd.to_numeric(
    amount_data["is_fraud"], errors="coerce"
)

if not amount_data["is_fraud"].isin([0, 1]).all():
    raise ValueError("is_fraud contains missing or unexpected values.")

valid_amount = (
    np.isfinite(amount_data["transaction_amount"])
    & amount_data["transaction_amount"].ge(0)
)

print("Rows excluded from amount analysis:", int((~valid_amount).sum()))
print(
    "Zero-value transactions:",
    int(amount_data["transaction_amount"].eq(0).sum()),
)

amount_data = amount_data.loc[valid_amount].copy()

if amount_data.empty:
    raise ValueError("No valid transaction amounts available.")

amount_data["transaction_type"] = amount_data["is_fraud"].map(
    {0: "Legitimate", 1: "Fraud"}
)

amount_summary = (
    amount_data.groupby("transaction_type")["transaction_amount"]
    .agg(
        transaction_count="size",
        total_transaction_amount="sum",
        mean_amount="mean",
        median_amount="median",
        p90_amount=lambda values: values.quantile(0.90),
        p99_amount=lambda values: values.quantile(0.99),
        maximum_amount="max",
    )
    .reindex(["Legitimate", "Fraud"])
)

display(amount_summary.round(2))

amount_data["log_amount"] = np.log1p(
    amount_data["transaction_amount"]
)

bin_edges = np.histogram_bin_edges(
    amount_data["log_amount"], bins=40
)

colors = {"Legitimate": "#25EB32", "Fraud": "#DC2626"}

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for transaction_type, color in colors.items():
    values = amount_data.loc[
        amount_data["transaction_type"].eq(transaction_type),
        "log_amount",
    ]

    if not values.empty:
        axes[0].hist(
            values,
            bins=bin_edges,
            weights=np.full(len(values), 100 / len(values)),
            histtype="step",
            linewidth=2,
            color=color,
            label=transaction_type,
        )

axes[0].set_title("Amount Distribution Within Each Class")
axes[0].set_xlabel("log(1 + transaction amount)")
axes[0].set_ylabel("Percentage of class transactions per bin")
axes[0].legend()

amount_summary[["mean_amount", "median_amount"]].plot.bar(
    ax=axes[1],
    color=["#475569", "#F59E0B"],
    rot=0,
)

axes[1].set_title("Mean vs Median Transaction Amount")
axes[1].set_xlabel("")
axes[1].set_ylabel("Amount (dataset units)")
axes[1].legend(["Mean", "Median"])

for ax in axes:
    ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

hour_data = df[["transaction_datetime", "is_fraud"]].copy()

hour_data["transaction_datetime"] = pd.to_datetime(
    hour_data["transaction_datetime"],
    errors="coerce",
)

hour_data["is_fraud"] = pd.to_numeric(
    hour_data["is_fraud"],
    errors="coerce",
)

if not hour_data["is_fraud"].isin([0, 1]).all():
    raise ValueError("is_fraud must contain only 0 and 1.")

invalid_dates = hour_data["transaction_datetime"].isna()
print("Rows excluded due to missing/invalid dates:", int(invalid_dates.sum()))

hour_data = hour_data.loc[~invalid_dates].copy()

if hour_data.empty:
    raise ValueError("No valid timestamps available for hourly analysis.")

hour_data["hour"] = hour_data["transaction_datetime"].dt.hour

hour_summary = (
    hour_data.groupby("hour")
    .agg(
        total_transactions=("is_fraud", "size"),
        fraud_transactions=("is_fraud", "sum"),
    )
    .reindex(range(24), fill_value=0)
)

hour_summary["fraud_rate_pct"] = (
    hour_summary["fraud_transactions"]
    .div(hour_summary["total_transactions"].where(
        hour_summary["total_transactions"].gt(0)
    ))
    .mul(100)
)

overall_fraud_rate = hour_data["is_fraud"].mean() * 100

display(
    hour_summary.style.format({
        "total_transactions": "{:,.0f}",
        "fraud_transactions": "{:,.0f}",
        "fraud_rate_pct": "{:.2f}%",
    }, na_rep="No transactions")
)

fig, axes = plt.subplots(2, 1, figsize=(8, 6), sharex=True)

axes[0].bar(
    hour_summary.index,
    hour_summary["fraud_transactions"],
    color="#DC2626",
)
axes[0].set_title("Fraud Transaction Count by Hour")
axes[0].set_ylabel("Fraud transactions")

axes[1].plot(
    hour_summary.index,
    hour_summary["fraud_rate_pct"],
    color="#2563EB",
    marker="o",
    linewidth=2,
    label="Hourly fraud rate",
)
axes[1].axhline(
    overall_fraud_rate,
    color="#64748B",
    linestyle="--",
    label=f"Overall fraud rate: {overall_fraud_rate:.2f}%",
)

axes[1].set_title("Fraud Rate by Hour")
axes[1].set_xlabel("Hour of day (dataset time; timezone unverified)")
axes[1].set_ylabel("Fraud rate (%)")
axes[1].set_xticks(range(24))
axes[1].legend()

for ax in axes:
    ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

merchant_data = df[[
    "merchant_category",
    "transaction_amount",
    "is_fraud",
]].copy()

merchant_data["is_fraud"] = pd.to_numeric(
    merchant_data["is_fraud"], errors="coerce"
)

merchant_data["transaction_amount"] = pd.to_numeric(
    merchant_data["transaction_amount"], errors="coerce"
)

if not merchant_data["is_fraud"].isin([0, 1]).all():
    raise ValueError("is_fraud must contain only 0 and 1.")

categories = merchant_data["merchant_category"].astype("string").str.strip()

unknown_category = (
    categories.isna()
    | categories.str.lower().isin(["", "n/a", "na", "null", "none", "nan"])
)

merchant_data["merchant_category"] = categories.mask(
    unknown_category, "Unknown"
)

valid_amount = (
    np.isfinite(merchant_data["transaction_amount"])
    & merchant_data["transaction_amount"].ge(0)
)

merchant_data["valid_amount"] = (
    merchant_data["transaction_amount"].where(valid_amount)
)

merchant_data["fraud_transaction_amount"] = (
    merchant_data["valid_amount"].where(
        merchant_data["is_fraud"].eq(1)
    )
)

print("Transactions with invalid amounts:", int((~valid_amount).sum()))

merchant_summary = (
    merchant_data.groupby("merchant_category")
    .agg(
        total_transactions=("is_fraud", "size"),
        fraud_transactions=("is_fraud", "sum"),
        valid_amount_count=("valid_amount", "count"),
        total_transaction_amount=(
            "valid_amount", lambda values: values.sum(min_count=1)
        ),
        fraud_transaction_amount=(
            "fraud_transaction_amount",
            lambda values: values.sum(min_count=1),
        ),
    )
)

merchant_summary.loc[
    merchant_summary["fraud_transactions"].eq(0),
    "fraud_transaction_amount",
] = 0

merchant_summary["fraud_rate_pct"] = (
    merchant_summary["fraud_transactions"]
    / merchant_summary["total_transactions"]
    * 100
)

merchant_summary = merchant_summary.sort_values(
    "fraud_rate_pct", ascending=False
)

display(merchant_summary.round(2))

plot_data = merchant_summary.head(10).sort_values("fraud_rate_pct")
overall_fraud_rate = merchant_data["is_fraud"].mean() * 100

fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

axes[0].barh(
    plot_data.index,
    plot_data["fraud_rate_pct"],
    color="#DC2626",
)
axes[0].axvline(
    overall_fraud_rate,
    color="#64748B",
    linestyle="--",
    label=f"Overall: {overall_fraud_rate:.2f}%",
)
axes[0].set_title("Top 10 Categories by Observed Fraud Rate")
axes[0].set_xlabel("Fraud rate (%)")
axes[0].legend()

axes[1].barh(
    plot_data.index,
    plot_data["total_transactions"],
    color="#2563EB",
)
axes[1].set_title("Transaction Volume for These Categories")
axes[1].set_xlabel("Total transactions")

for ax in axes:
    ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

frequency_data = df[["transactions_last_1h", "is_fraud"]].copy()

frequency_data["transactions_last_1h"] = pd.to_numeric(
    frequency_data["transactions_last_1h"],
    errors="coerce",
)
frequency_data["is_fraud"] = pd.to_numeric(
    frequency_data["is_fraud"],
    errors="coerce",
)

if not frequency_data["is_fraud"].isin([0, 1]).all():
    raise ValueError("is_fraud must contain only 0 and 1.")

transaction_counts = frequency_data["transactions_last_1h"]

valid_count = (
    np.isfinite(transaction_counts)
    & transaction_counts.ge(0)
    & transaction_counts.mod(1).eq(0)
)

print("Rows excluded due to invalid counts:", int((~valid_count).sum()))

frequency_data = frequency_data.loc[valid_count].copy()

if frequency_data.empty:
    raise ValueError("No valid transaction-frequency values available.")

frequency_labels = ["0", "1", "2–3", "4–5", "6–10", "11+"]

frequency_data["frequency_group"] = pd.cut(
    frequency_data["transactions_last_1h"],
    bins=[0, 1, 2, 4, 6, 11, np.inf],
    labels=frequency_labels,
    right=False,
)

frequency_summary = (
    frequency_data.groupby("frequency_group", observed=False)
    .agg(
        total_transactions=("is_fraud", "size"),
        fraud_transactions=("is_fraud", "sum"),
    )
)

frequency_summary["fraud_rate_pct"] = (
    frequency_summary["fraud_transactions"]
    .div(
        frequency_summary["total_transactions"].where(
            frequency_summary["total_transactions"].gt(0)
        )
    )
    .mul(100)
)

overall_rate = frequency_data["is_fraud"].mean() * 100

frequency_summary["fraud_rate_lift"] = (
    frequency_summary["fraud_rate_pct"] / overall_rate
    if overall_rate > 0
    else np.nan
)

display(
    frequency_summary.style.format({
        "total_transactions": "{:,.0f}",
        "fraud_transactions": "{:,.0f}",
        "fraud_rate_pct": "{:.2f}%",
        "fraud_rate_lift": "{:.2f}x",
    }, na_rep="Not available")
)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].bar(
    frequency_labels,
    frequency_summary["total_transactions"],
    color="#2563EB",
)
axes[0].set_title("Transaction Volume by Last-Hour Frequency")
axes[0].set_ylabel("Number of transaction records")

axes[1].bar(
    frequency_labels,
    frequency_summary["fraud_rate_pct"],
    color="#DC2626",
)
axes[1].axhline(
    overall_rate,
    color="#64748B",
    linestyle="--",
    label=f"Baseline: {overall_rate:.2f}%",
)
axes[1].set_title("Fraud Rate by Last-Hour Frequency")
axes[1].set_ylabel("Fraud rate (%)")
axes[1].legend()

for ax in axes:
    ax.set_xlabel("Reported transactions in the last 1 hour")
    ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter

# Prepare an independent analysis dataset.
monthly_data = df[["transaction_datetime", "is_fraud"]].copy()

monthly_data["transaction_datetime"] = pd.to_datetime(
    monthly_data["transaction_datetime"],
    errors="coerce",
)

monthly_data["fraud_flag"] = monthly_data["is_fraud"].map({
    "No": 0,
    "Yes": 1,
})

if monthly_data["fraud_flag"].isna().any():
    raise ValueError("is_fraud contains missing or unexpected labels.")

invalid_dates = monthly_data["transaction_datetime"].isna()
print("Rows excluded due to invalid dates:", int(invalid_dates.sum()))

monthly_data = monthly_data.loc[~invalid_dates].copy()

if monthly_data.empty:
    raise ValueError("No valid timestamps available.")

monthly_data["month"] = (
    monthly_data["transaction_datetime"].dt.to_period("M")
)

monthly_summary = (
    monthly_data.groupby("month")
    .agg(
        total_transactions=("fraud_flag", "size"),
        fraud_transactions=("fraud_flag", "sum"),
    )
    .sort_index()
)

month_range = pd.period_range(
    monthly_data["month"].min(),
    monthly_data["month"].max(),
    freq="M",
)

monthly_summary = monthly_summary.reindex(
    month_range,
    fill_value=0,
)
monthly_summary.index.name = "month"

monthly_summary["fraud_rate_pct"] = (
    monthly_summary["fraud_transactions"]
    .div(
        monthly_summary["total_transactions"].where(
            monthly_summary["total_transactions"].gt(0)
        )
    )
    .mul(100)
)

display(
    monthly_summary.style.format({
        "total_transactions": "{:,.0f}",
        "fraud_transactions": "{:,.0f}",
        "fraud_rate_pct": "{:.2f}%",
    }, na_rep="No transactions")
)

month_labels = monthly_summary.index.strftime("%b %Y")
positions = range(len(monthly_summary))
overall_rate = monthly_data["fraud_flag"].mean() * 100

fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

axes[0].bar(
    positions,
    monthly_summary["total_transactions"],
    color="#01EEA7",
)
axes[0].set_title("Monthly Transaction Volume")
axes[0].set_ylabel("Total transactions")

axes[1].plot(
    positions,
    monthly_summary["fraud_rate_pct"],
    color="#DC2626",
    marker="o",
    linewidth=2,
    label="Monthly fraud rate",
)
axes[1].axhline(
    overall_rate,
    color="#64748B",
    linestyle="--",
    label=f"Overall: {overall_rate:.2f}%",
)
axes[1].set_title("Monthly Fraud Rate")
axes[1].set_ylabel("Fraud rate")
axes[1].yaxis.set_major_formatter(PercentFormatter(xmax=100))
axes[1].set_ylim(bottom=0)
axes[1].set_xticks(list(positions))
axes[1].set_xticklabels(month_labels, rotation=45, ha="right")
axes[1].legend()

for ax in axes:
    ax.spines[["top", "right"]].set_visible(False)

fig.suptitle("Monthly Trends — Adjusted Dataset Timeline", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

model_data = df.copy()

model_data["transaction_datetime"] = pd.to_datetime(
    model_data["transaction_datetime"],
    errors="raise",
)

model_data["is_fraud"] = model_data["is_fraud"].map({
    "No": 0,
    "Yes": 1,
})

if model_data["transaction_datetime"].isna().any():
    raise ValueError("Missing timestamps must be resolved before splitting.")

if model_data["is_fraud"].isna().any():
    raise ValueError("Unexpected or missing fraud labels.")

model_data = model_data.sort_values(
    "transaction_datetime",
    kind="stable",
).reset_index(drop=True)

if len(model_data) < 3:
    raise ValueError("Not enough rows for three chronological sets.")

train_cutoff = model_data.loc[
    int(len(model_data) * 0.70), "transaction_datetime"
]
test_cutoff = model_data.loc[
    int(len(model_data) * 0.85), "transaction_datetime"
]

timestamps = model_data["transaction_datetime"]

train_df = model_data.loc[timestamps < train_cutoff].copy()

validation_df = model_data.loc[
    (timestamps >= train_cutoff) & (timestamps < test_cutoff)
].copy()

test_df = model_data.loc[timestamps >= test_cutoff].copy()

datasets = {
    "Train": train_df,
    "Validation": validation_df,
    "Test": test_df,
}

if any(dataset.empty for dataset in datasets.values()):
    raise ValueError("A split is empty; timestamp boundaries need review.")

split_summary = pd.DataFrame([
    {
        "dataset": name,
        "rows": len(dataset),
        "start_date": dataset["transaction_datetime"].min(),
        "end_date": dataset["transaction_datetime"].max(),
    }
    for name, dataset in datasets.items()
])

display(split_summary)
```

```python
excluded_columns = {
    "is_fraud",
    "fraud_type",
    "fraud_amount_lost",
    "transaction_id",
    "customer_id",
    "card_id",
    "merchant_id",
    "transaction_datetime",
}

excluded_columns.update({
    "transactions_last_1h",
    "transactions_last_24h",
    "transactions_last_7d",
    "amount_last_1h",
    "amount_last_24h",
    "is_weekend",
    "num_chargebacks_ever",
})

feature_columns = [
    column
    for column in train_df.columns
    if column not in excluded_columns
    and not column.endswith("_original")
]

if not feature_columns:
    raise ValueError("No model features remain.")

def prepare_features_and_target(data, feature_columns):
    """Separate candidate features from the validated binary target."""
    required_columns = set(feature_columns) | {"is_fraud"}
    missing_columns = required_columns.difference(data.columns)

    if missing_columns:
        raise ValueError(f"Missing columns: {sorted(missing_columns)}")

    if not data["is_fraud"].isin([0, 1]).all():
        raise ValueError("Expected numeric target values: 0 and 1.")

    X = data.loc[:, feature_columns].copy()
    y = data["is_fraud"].astype("int64").copy()

    return X, y


X_train, y_train = prepare_features_and_target(
    train_df, feature_columns
)
X_validation, y_validation = prepare_features_and_target(
    validation_df, feature_columns
)
X_test, y_test = prepare_features_and_target(
    test_df, feature_columns
)

print("Training features:", X_train.shape)
print("Validation features:", X_validation.shape)
print("Test features:", X_test.shape)

print("\nCandidate input features:")
for number, column in enumerate(feature_columns, start=1):
    print(f"{number:02d}. {column}")
```

```python
import numpy as np
import pandas as pd

# Profile training data only.
feature_audit = pd.DataFrame({
    "data_type": X_train.dtypes.astype(str),
    "missing_count": X_train.isna().sum(),
    "unique_values": X_train.nunique(dropna=True),
})

feature_audit["blank_or_placeholder_count"] = 0
feature_audit["infinite_count"] = 0

for column in X_train.columns:
    values = X_train[column]

    if pd.api.types.is_numeric_dtype(values):
        feature_audit.loc[column, "infinite_count"] = int(
            values.isin([np.inf, -np.inf]).sum()
        )
    else:
        normalized = values.astype("string").str.strip().str.lower()

        feature_audit.loc[column, "blank_or_placeholder_count"] = int(
            normalized.isin(["", "n/a", "na", "null", "none", "nan"]).sum()
        )

feature_audit.index.name = "feature"

# Print every row without notebook table truncation.
print(feature_audit.to_string())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_features = X_train.select_dtypes(
    include="number"
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "string", "category", "bool"]
).columns.tolist()

unsupported_features = set(X_train.columns) - set(
    numeric_features + categorical_features
)

if unsupported_features:
    raise ValueError(
        f"Unsupported feature types: {sorted(unsupported_features)}"
    )

numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])

categorical_pipeline = Pipeline(steps=[
    (
        "imputer",
        SimpleImputer(
            strategy="constant",
            fill_value="Missing",
        ),
    ),
    (
        "encoder",
        OneHotEncoder(handle_unknown="ignore"),
    ),
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features),
    ],
    remainder="drop",
)

print("Numeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
print("Preprocessing pipeline created successfully.")
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

logistic_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    (
        "classifier",
        LogisticRegression(
            class_weight="balanced",
            solver="lbfgs",
            max_iter=2000,
            random_state=42,
        ),
    ),
])

logistic_pipeline.fit(X_train, y_train)

print("Logistic Regression training completed.")

fraud_class_index = list(
    logistic_pipeline.named_steps["classifier"].classes_
).index(1)

validation_scores = logistic_pipeline.predict_proba(
    X_validation
)[:, fraud_class_index]

print("Validation scores generated:", len(validation_scores))
```

```python
import numpy as np
import pandas as pd

def evaluate_review_capacity(y_true, scores, capacities):
    """Evaluate the highest-scored transactions at fixed review budgets."""
    labels = np.asarray(y_true)
    scores = np.asarray(scores, dtype=float)

    if labels.ndim != 1 or scores.ndim != 1:
        raise ValueError("Labels and scores must be one-dimensional.")

    if len(labels) != len(scores) or len(labels) == 0:
        raise ValueError("Labels and scores must have equal, non-zero length.")

    if not np.isin(labels, [0, 1]).all() or not np.isfinite(scores).all():
        raise ValueError("Invalid labels or scores.")

    # Stable ordering breaks equal-score ties using existing row order.
    ranked_positions = np.argsort(-scores, kind="stable")
    total_fraud = int(labels.sum())
    results = []

    for capacity in capacities:
        if not 1 <= capacity <= len(labels):
            raise ValueError(f"Invalid review capacity: {capacity}")

        selected_labels = labels[ranked_positions[:capacity]]
        detected = int(selected_labels.sum())

        results.append({
            "review_capacity": capacity,
            "fraud_detected": detected,
            "false_alerts": capacity - detected,
            "fraud_missed": total_fraud - detected,
            "precision_at_k": detected / capacity,
            "recall_at_k": (
                detected / total_fraud if total_fraud else np.nan
            ),
        })

    return pd.DataFrame(results)


capacity_reports = []

for model_name, scores in {
    "Logistic Regression": validation_scores,
    "Random Forest": rf_validation_scores,
}.items():
    report = evaluate_review_capacity(
        y_validation,
        scores,
        capacities=[100, 250, 500, 1000],
    )
    report.insert(0, "model", model_name)
    capacity_reports.append(report)

capacity_comparison = (
    pd.concat(capacity_reports, ignore_index=True)
    .sort_values(["review_capacity", "model"])
)

display(
    capacity_comparison.style.format({
        "precision_at_k": "{:.2%}",
        "recall_at_k": "{:.2%}",
    }, na_rep="Not defined")
)
```

```python
from pathlib import Path
import pandas as pd

sql_columns = [
    "transaction_id",
    "customer_id",
    "merchant_id",
    "transaction_datetime",
    "transaction_amount",
    "merchant_category",
    "merchant_country",
    "card_present",
    "customer_age",
    "account_age_days",
    "is_fraud",
]

sql_data = df.loc[:, sql_columns].copy()

# Convert supported fraud labels to PostgreSQL boolean text.
labels = sql_data["is_fraud"].astype("string").str.strip().str.lower()

sql_data["is_fraud"] = labels.map({
    "yes": "true", "1": "true", "1.0": "true", "true": "true",
    "no": "false", "0": "false", "0.0": "false", "false": "false",
})

if sql_data["is_fraud"].isna().any():
    raise ValueError("Missing or unexpected fraud labels found.")

sql_data["transaction_datetime"] = pd.to_datetime(
    sql_data["transaction_datetime"],
    errors="raise",
)

required_columns = [
    "transaction_id", "customer_id", "merchant_id",
    "transaction_datetime", "transaction_amount", "is_fraud",
]

if sql_data[required_columns].isna().any().any():
    raise ValueError("Required database fields contain missing values.")

if sql_data["transaction_id"].duplicated().any():
    raise ValueError("Duplicate transaction IDs found.")

# Ensure integer columns export without decimal suffixes.
for column in ["card_present", "customer_age", "account_age_days"]:
    sql_data[column] = pd.to_numeric(
        sql_data[column], errors="raise"
    ).astype("Int64")

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

export_dir = project_root / "data" / "processed"
export_dir.mkdir(parents=True, exist_ok=True)

csv_path = export_dir / "transactions_postgresql.csv"

sql_data.to_csv(
    csv_path,
    index=False,
    encoding="utf-8",
    date_format="%Y-%m-%d %H:%M:%S",
)

print("Exported rows:", len(sql_data))
print("CSV location:", csv_path.resolve())
```
